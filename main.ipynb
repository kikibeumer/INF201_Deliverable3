## Class overview

### Dice
An n-sided die used to introduce randomness

**Attributes**
* `sides: int`: Number of sides (default 6, must be >=2, otherwise `ValueError`)

**Methods**
* `roll()`: Returns a random integer from 1 to `sides``

* `__call__()`: Lets you write `d20()` instead of `d20.roll()`

* `__repr__()`: Returns e.g. `Dice(sides=6)`



### Weapon
An item a character can equip. Damage comes from a `Dice`object

**Attributes**
* `name: str`: Name of the weapon

* `damage_dice: Dice`: The die rolled to determine damage

**Methods**
* `roll_damage()`: Rolls `damage_dice` and returns the damage as an int

### Character
The base class for all fighters

**Attributes**
* `name: str`: Character's name

* `_hp: int`: Current HP (protected, accessed via the `hp` property)

* `hp_max: int`: Maximum HP (default 100)

* `damage: int`: Base damage without a weapon (default 10)

* `weapon: Weapon = None`: Optional equipped weapon

**Methods**
* `attack(target)`: Rolls weapon damage if equipped, otherwise uses `damage`, then calls `target.take_damage()``

* `take_damage(damage)`: Reduces own HP, never below 0

* `heal(heal)`: Increases own HP, never above `hp_max`

* `introduce()`: Prints a greeting

* `hp` (property): Read-only access to `_hp``

* `is_alive` (property): `True` if `_hp > 0`

* `__repr__()` & `__str__()`: Developer and readable string representations

### Warrior(Character)
Subclass of Character. A defensive specialist.

**Added attributes**
* `armor: int`: Armor rating (default 5)

**Overridden methods**
* `take_damage(damage)`: Reduces incoming damage by `armor` (minimum 0) before applying it. 

### Mage(Character)
Subclass of Character. A resource-based attacker. 

**Added attributes**
* `mana: int`: Current mana (default 30)

* `mana_max: int`: Maximum mana (default 30)

**Overridden methods**
* `attack(target)`: With >=10 mana: spends 10 mana and deals double damage, ignoring weapons. Otherwise: base attack with the equipped weapon and restores 5 mana

### Battle
Manages turn-based combat between two characters.

**Attributes**
* `hero: Character`: The hero

* `enemy: Character`: The enemy

**Methods**
* `play_round()`: Hero attacks; if the enemy is still alive, the enemy attacks back

* `simulate_battle()`: Repeats rounds until one character is no longer alive
